# Global Polymeric Materials Knowledge Graph
This notebook builds a graph database to identify technology clusters, emerging materials, and innovation bottlenecks.

```python
import networkx as nx
import matplotlib.pyplot as plt

# Initialize the directed graph
G = nx.DiGraph()

# Define node categories
categories = {
    'Polymers': ['PLA', 'PHB', 'PEEK', 'Graphene-Reinforced Epoxy'],
    'Additives': ['Carbon Nanotubes', 'Plasticizers', 'UV Stabilizers'],
    'Manufacturing': ['3D Printing', 'Injection Molding', 'Electrospinning'],
    'Applications': ['Aerospace', 'Biomedical Implants', 'Sustainable Packaging'],
    'Patents': ['US1234567', 'EP9876543']
}

# Add nodes with attributes
for category, names in categories.items():
    for name in names:
        G.add_node(name, category=category)

# Add example edges based on the prompt relationships
# [Source, Target, Relationship]
relationships = [
    ('Carbon Nanotubes', 'Graphene-Reinforced Epoxy', 'Improves'),
    ('PLA', 'Sustainable Packaging', 'Used with'),
    ('3D Printing', 'PEEK', 'Manufactured by'),
    ('Graphene-Reinforced Epoxy', 'Aerospace', 'Used with'),
    ('PHB', 'PLA', 'Replaces'),
    ('US1234567', 'Graphene-Reinforced Epoxy', 'Analysis')
]

for u, v, rel in relationships:
    G.add_edge(u, v, label=rel)

print(f"Graph created with {G.number_of_nodes()} nodes and {G.number_of_edges()} edges.")
```

```python
# Simple visualization of the Knowledge Graph
plt.figure(figsize=(12, 8))
pos = nx.spring_layout(G, seed=42)

# Color map based on categories
color_map = {'Polymers': 'skyblue', 'Additives': 'lightgreen', 'Manufacturing': 'orange', 'Applications': 'pink', 'Patents': 'grey'}
node_colors = [color_map[G.nodes[n]['category']] for n in G.nodes]

nx.draw(G, pos, with_labels=True, node_color=node_colors, node_size=3000, font_size=10, font_weight='bold', edge_color='gray', arrows=True)
edge_labels = nx.get_edge_attributes(G, 'label')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.title("Polymeric Materials Knowledge Graph Subset")
plt.show()
```

```python
import networkx.algorithms.community as nx_comm
import networkx as nx

def identify_clusters(graph):
    communities = nx_comm.greedy_modularity_communities(graph.to_undirected())
    print("--- Technology Clusters ---")
    for i, comm in enumerate(communities):
        print(f"Cluster {i+1}: {list(comm)}")

def identify_emerging_materials(graph):
    centrality = nx.degree_centrality(graph)
    sorted_centrality = sorted(centrality.items(), key=lambda x: x[1], reverse=True)
    print("\n--- Top Influential/Emerging Nodes ---")
    for node, score in sorted_centrality[:5]:
        print(f"{node}: {score:.2f}")

def identify_bottlenecks(graph):
    betweenness = nx.betweenness_centrality(graph)
    sorted_betweenness = sorted(betweenness.items(), key=lambda x: x[1], reverse=True)
    print("\n--- Potential Innovation Bottlenecks (Bridges) ---")
    for node, score in sorted_betweenness[:3]:
        print(f"{node}: {score:.2f}")

if 'G' in globals():
    identify_clusters(G)
    identify_emerging_materials(G)
    identify_bottlenecks(G)
else:
    print("Error: Graph 'G' is not defined. Please run the initialization cell (e29a2282) first.")
```
