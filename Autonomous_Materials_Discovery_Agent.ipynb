## Implementation: Research and Database Agents

We will use `langchain` to create a research agent that can browse the web or search papers, and a database agent to fetch structured material properties.

```python
from langchain_openai import ChatOpenAI
from langchain.prompts import ChatPromptTemplate

# Initialize the LLM
llm = ChatOpenAI(model="gpt-4o", temperature=0)

def research_agent_run(material_topic):
    """Uses LLM to summarize current research on a material class."""
    prompt = ChatPromptTemplate.from_template("Summarize recent breakthroughs in {topic} for energy applications.")
    chain = prompt | llm
    return chain.invoke({"topic": material_topic}).content

# Example usage (uncomment after adding API key):
# print(research_agent_run("solid-state battery electrolytes"))
```
