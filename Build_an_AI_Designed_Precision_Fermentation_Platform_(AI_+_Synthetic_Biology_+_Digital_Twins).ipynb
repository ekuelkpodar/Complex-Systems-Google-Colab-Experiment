## Phase 2: Metabolic Digital Twins & FBA Optimization
In this section, we model the internal metabolic state of the cell. We simulate carbon flux and visualize the network of reactions that transform raw nutrients into our target bio-products.

```python
import networkx as nx
from pyvis.network import Network
import IPython

def generate_metabolic_graph():
    G = nx.DiGraph()

    # Define nodes: Metabolites (blue) and Reactions (orange)
    nodes = [
        ("Glucose", {"color": "#2196F3", "size": 25}),
        ("G6P", {"color": "#2196F3", "size": 20}),
        ("Pyruvate", {"color": "#2196F3", "size": 20}),
        ("Acetyl-CoA", {"color": "#2196F3", "size": 20}),
        ("TCA Cycle", {"color": "#FF9800", "size": 30}),
        ("Target Protein", {"color": "#E91E63", "size": 35}),
        ("CO2", {"color": "#9E9E9E", "size": 15})
    ]

    G.add_nodes_from(nodes)

    # Define pathways (edges)
    edges = [
        ("Glucose", "G6P", "Hexokinase"),
        ("G6P", "Pyruvate", "Glycolysis"),
        ("Pyruvate", "Acetyl-CoA", "PDH Complex"),
        ("Acetyl-CoA", "TCA Cycle", "Citrate Synthase"),
        ("TCA Cycle", "Target Protein", "Anabolism"),
        ("TCA Cycle", "CO2", "Respiration")
    ]

    for u, v, label in edges:
        G.add_edge(u, v, title=label, label=label, arrows="to")

    return G

# Visualize with PyVis for an interactive experience
metabolic_net = Network(height="500px", width="100%", bgcolor="#121212", font_color="white", notebook=True, cdn_resources='remote')
metabolic_net.from_nx(generate_metabolic_graph())
metabolic_net.show("metabolic_twin.html")
IPython.display.HTML(filename="metabolic_twin.html")
```

## Phase 3: Bioreactor Digital Twin & Process Simulation
This module simulates the physical environment of the fermentation tank, modeling cell growth ($X$), substrate consumption ($S$), and product formation ($P$) over time using differential equations.

```python
def simulate_fermentation(hours=72):
    t = np.linspace(0, hours, 100)
    # Kinetic parameters
    mu_max = 0.35
    Ks = 0.5
    Y_xs = 0.5
    Y_px = 0.2

    X = [0.1]
    S = [100.0]
    P = [0.0]
    dt = t[1] - t[0]

    for i in range(len(t)-1):
        mu = mu_max * S[-1] / (Ks + S[-1])
        dX = mu * X[-1]
        dS = -(1/Y_xs) * dX
        dP = Y_px * dX

        X.append(max(0, X[-1] + dX * dt))
        S.append(max(0, S[-1] + dS * dt))
        P.append(P[-1] + dP * dt)

    return t, X, S, P

t, biomass, substrate, product = simulate_fermentation()

fig = go.Figure()
fig.add_trace(go.Scatter(x=t, y=biomass, name="Biomass (OD600)", line=dict(color='#00e676')))
fig.add_trace(go.Scatter(x=t, y=substrate, name="Glucose (g/L)", line=dict(color='#2196F3')))
fig.add_trace(go.Scatter(x=t, y=product, name="Target Protein (g/L)", line=dict(color='#E91E63')))

fig.update_layout(
    title="Digital Twin: Real-Time Bioproduct Kinetic Simulation",
    xaxis_title="Time (Hours)",
    yaxis_title="Concentration",
    template="plotly_dark",
    hovermode="x unified"
)
fig.show()
```

## Phase 5: World Model & Strategic 'What-If' Engine
This module simulates global market dynamics and supply chain disruptions to stress-test the production infrastructure.

```python
def simulate_market_shock(scenario="Protein Demand Surge"):
    years = np.arange(2024, 2036)
    base_demand = 100 * (1.05 ** (years - 2024))

    if scenario == "Protein Demand Surge":
        shock_demand = base_demand.copy()
        shock_demand[7:] *= 5 # 500% increase by 2031
        title = "Scenario: 500% Alternative Protein Demand Explosion (2035)"

    fig = go.Figure()
    fig.add_trace(go.Scatter(x=years, y=base_demand, name="Base Case", fill='tozeroy'))
    fig.add_trace(go.Scatter(x=years, y=shock_demand, name="AI-Predicted Surge", line=dict(width=4, dash='dot')))

    fig.update_layout(title=title, xaxis_title="Year", yaxis_title="Demand Index", template="plotly_dark")
    fig.show()

simulate_market_shock()
```
