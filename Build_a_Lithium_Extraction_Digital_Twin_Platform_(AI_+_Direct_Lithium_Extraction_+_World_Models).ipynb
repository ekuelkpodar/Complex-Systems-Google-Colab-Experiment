## Synthetic Brine Operation and Reservoir Simulation

### Subtask:
Develop a numerical simulation to model lithium extraction rates and reservoir depletion over time based on the geochemistry data in df_brine_chemistry.


**Reasoning**:
I will define the simulation parameters and create a function to generate the production data for each region, then store it in a long-format DataFrame for analysis.



```python
import pandas as pd
import numpy as np
import plotly.express as px

# Simulation Parameters
months = 120
flow_rate_m3_month = 50000 # Constant brine extraction rate
depletion_constant = 0.005 # Rate at which concentration declines per month

sim_results = []

for index, row in df_brine_chemistry.iterrows():
    region = row['Region']
    initial_conc = row['Li_Concentration_mgL']

    current_conc = initial_conc
    for month in range(1, months + 1):
        # Production (kg) = Flow (m3) * Conc (mg/L) / 1000
        # (Note: 1 mg/L = 1 g/m3, so mg/L * m3 / 1000 = kg)
        production_kg = (flow_rate_m3_month * current_conc) / 1000

        sim_results.append({
            'Region': region,
            'Month': month,
            'Lithium_Production_kg': production_kg,
            'Concentration_mgL': current_conc
        })

        # Model exponential depletion of the resource
        current_conc *= (1 - depletion_constant)

df_reservoir_simulation = pd.DataFrame(sim_results)
print("Reservoir simulation complete. First 5 rows:")
print(df_reservoir_simulation.head())

# Visualize the production decline
fig_sim = px.line(
    df_reservoir_simulation,
    x='Month',
    y='Lithium_Production_kg',
    color='Region',
    title='10-Year Simulated Lithium Production Decline Curves',
    labels={'Lithium_Production_kg': 'Monthly Production (kg)', 'Month': 'Month of Operation'},
    template='plotly_dark'
)
fig_sim.show()
```

## Chemical DLE Process Modeling

### Subtask:
Model the Direct Lithium Extraction (DLE) efficiency and chemical reagent requirements based on brine chemistry.


**Reasoning**:
I will define a DLE efficiency model that accounts for the Mg/Li ratio interference and calculate the effective recovery for each region.



```python
import pandas as pd
import numpy as np

def model_dle_performance(row):
    # Higher Mg/Li ratio typically decreases efficiency and increases reagent cost
    # Base efficiency is 90%, but drops as Mg/Li ratio increases
    efficiency = max(0.6, 0.9 - (row['Mg_Li_Ratio'] * 0.001))

    # Reagent consumption (kg of reagent per kg of Li produced)
    # Scales with Mg concentration
    reagent_intensity = 2.0 + (row['Mg_Li_Ratio'] * 0.05)

    return pd.Series([efficiency, reagent_intensity], index=['DLE_Efficiency', 'Reagent_Intensity_kg_per_kg'])

# Apply model to base brine data
df_dle_performance = df_brine_chemistry.apply(model_dle_performance, axis=1)
df_brine_performance = pd.concat([df_brine_chemistry, df_dle_performance], axis=1)

# Integrate with reservoir simulation to get 'Effective Production'
df_reservoir_simulation = df_reservoir_simulation.merge(df_brine_performance[['Region', 'DLE_Efficiency', 'Reagent_Intensity_kg_per_kg']], on='Region')
df_reservoir_simulation['Net_Lithium_Yield_kg'] = df_reservoir_simulation['Lithium_Production_kg'] * df_reservoir_simulation['DLE_Efficiency']
df_reservoir_simulation['Estimated_Reagent_Need_kg'] = df_reservoir_simulation['Net_Lithium_Yield_kg'] * df_reservoir_simulation['Reagent_Intensity_kg_per_kg']

print("DLE Process Modeling applied to simulation data.")
print(df_brine_performance[['Region', 'DLE_Efficiency', 'Reagent_Intensity_kg_per_kg']])
```

## Prepare Sankey Data

### Subtask:
Process simulation results into a mass-balance format suitable for a Sankey diagram.


**Reasoning**:
I will aggregate the reservoir simulation data to determine the total mass balance of lithium and define the nodes for the Sankey flow.



```python
import pandas as pd

# 1. Aggregate total flows per region
sankey_totals = df_reservoir_simulation.groupby('Region').agg({
    'Lithium_Production_kg': 'sum',
    'Net_Lithium_Yield_kg': 'sum'
}).reset_index()

sankey_totals['Lithium_Losses_kg'] = sankey_totals['Lithium_Production_kg'] - sankey_totals['Net_Lithium_Yield_kg']

# 2. Define Nodes
# Indices: 0-3 (Regions), 4 (DLE Extraction Process), 5 (Marketable Lithium), 6 (Waste/Loss)
regions = sankey_totals['Region'].tolist()
nodes = regions + ['DLE Extraction Process', 'Marketable Lithium Product', 'Extraction Losses']

# 3. Define Flows (Sources, Targets, Values)
sources = []
targets = []
values = []

# Flow from Regions to DLE Process
for i, region in enumerate(regions):
    sources.append(i)
    targets.append(4) # DLE Process index
    values.append(sankey_totals.loc[i, 'Lithium_Production_kg'])

# Flow from DLE Process to Yield and Losses
# Total from DLE to Yield
sources.append(4)
targets.append(5)
values.append(sankey_totals['Net_Lithium_Yield_kg'].sum())

# Total from DLE to Waste
sources.append(4)
targets.append(6)
values.append(sankey_totals['Lithium_Losses_kg'].sum())

print("Sankey nodes and flow links prepared.")
print(f"Nodes: {nodes}")
```

## Generate Regional Sankey Diagram

### Subtask:
Construct an interactive Plotly Sankey diagram visualizing the mass balance of lithium extraction.


**Reasoning**:
I will use the `go.Sankey` object from Plotly to visualize the flows from the lithium source regions through the DLE process to the final product or losses. This provides a clear overview of extraction efficiency.



```python
import plotly.graph_objects as go

# Define the Sankey diagram
fig_sankey = go.Figure(data=[go.Sankey(
    node = dict(
      pad = 15,
      thickness = 20,
      line = dict(color = "black", width = 0.5),
      label = nodes,
      color = "blue"
    ),
    link = dict(
      source = sources,
      target = targets,
      value = values,
      label = [f"{v:,.0f} kg" for v in values],
      color = "rgba(0, 204, 150, 0.4)" # Translucent teal for links
  ))])

# Update layout for a professional look
fig_sankey.update_layout(
    title_text="Global Lithium Extraction Mass Balance: From Source to Marketable Product",
    font_size=12,
    template='plotly_white',
    height=600
)

fig_sankey.show()
print("Interactive Sankey diagram generated.")
```
