# Historical Civilization Network
Building a temporal network to visualize the connections between civilizations, empires, and trade routes.

```python
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd

# Initialize the directed graph
G = nx.DiGraph()

print("Libraries imported and Graph initialized.")
```

```python
# Define historical entities and their connections
civilizations = [
    ("Roman Empire", "Han China", {"relation": "Trade (Silk Road)", "weight": 1}),
    ("Roman Empire", "Persia", {"relation": "War", "weight": 3}),
    ("Persia", "Han China", {"relation": "Trade", "weight": 2}),
    ("British Empire", "India", {"relation": "Colonialism", "weight": 5}),
    ("British Empire", "China", {"relation": "Trade/Conflict", "weight": 4})
]

# Add edges to the graph
G.add_edges_from(civilizations)

# Basic visualization
plt.figure(figsize=(10, 6))
pos = nx.spring_layout(G)
nx.draw(G, pos, with_labels=True, node_color='skyblue', node_size=2000, edge_color='gray', font_size=10, font_weight='bold')
edge_labels = nx.get_edge_attributes(G, 'relation')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)
plt.title("Historical Civilization Network Connections")
plt.show()
```
