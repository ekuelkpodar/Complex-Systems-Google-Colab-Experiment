# AI-Powered Quantum Materials Literature Map
This notebook aims to download research papers from arXiv and construct a knowledge graph connecting materials, phenomena, scientists, and institutions.

```python
!pip install -q arxiv networkx pyvis pandas
```

```python
import arxiv
import pandas as pd
import networkx as nx
from pyvis.network import Network

# Define the search query for Quantum Materials
search_query = '"quantum materials" OR "topological insulators" OR "high-Tc superconductivity"'

client = arxiv.Client()
search = arxiv.Search(
    query=search_query,
    max_results=50,
    sort_by=arxiv.SortCriterion.Relevance
)

papers = []
for result in client.results(search):
    papers.append({
        'title': result.title,
        'authors': [author.name for author in result.authors],
        'published': result.published,
        'summary': result.summary,
        'pdf_url': result.pdf_url,
        'categories': result.categories
    })

df_papers = pd.DataFrame(papers)
display(df_papers.head())
```

```python
### Graph Generation Results
The knowledge graph is now fully constructed. It visualizes the relationships between:
- **Papers** (Blue)
- **Authors** (Red)
- **Categories** (Yellow)
- **Materials** (Green)
- **Phenomena** (Orange)
```

```python
G = nx.Graph()

for _, row in df_papers.iterrows():
    paper_title = row['title']
    G.add_node(paper_title, label=paper_title, title=paper_title, color='#97c2fc', group='Paper')

    for author in row['authors']:
        G.add_node(author, label=author, title=f'Author: {author}', color='#fb7e81', group='Scientist')
        G.add_edge(paper_title, author)

    for cat in row['categories']:
        G.add_node(cat, label=cat, title=f'Category: {cat}', color='#ffff00', group='Field')
        G.add_edge(paper_title, cat)

net = Network(height='600px', width='100%', bgcolor='#222222', font_color='white', notebook=True, cdn_resources='remote')
net.from_nx(G)
net.show('quantum_map.html')
```

## Entity Extraction: Materials and Phenomena
We will use a predefined list of common quantum materials and phenomena to tag the papers and expand our graph.

```python
materials_keywords = ["Graphene", "Bi2Se3", "WTe2", "NbSe2", "Cuprates", "Iron-based superconductors"]
phenomena_keywords = ["Superconductivity", "Majorana fermions", "Quantum Hall Effect", "Weyl semimetals", "Skyrmions"]

def extract_entities(text, keywords):
    return [k for k in keywords if k.lower() in text.lower()]

df_papers["extracted_materials"] = df_papers["summary"].apply(lambda x: extract_entities(x, materials_keywords))
df_papers["extracted_phenomena"] = df_papers["summary"].apply(lambda x: extract_entities(x, phenomena_keywords))

# Update Graph with new entities
for _, row in df_papers.iterrows():
    for mat in row["extracted_materials"]:
        G.add_node(mat, label=mat, title=f"Material: {mat}", color="#90ee90", group="Material")
        G.add_edge(row["title"], mat)
    for phen in row["extracted_phenomena"]:
        G.add_node(phen, label=phen, title=f"Phenomenon: {phen}", color="#ffa500", group="Phenomena")
        G.add_edge(row["title"], phen)

# Re-render updated graph
net_updated = Network(height="600px", width="100%", bgcolor="#222222", font_color="white", notebook=True, cdn_resources="remote")
net_updated.from_nx(G)
net_updated.show("quantum_map_enriched.html")
```

## Final Enrichment: Extracting Institutions
We'll use a regular expression to find common institution patterns (e.g., 'University of...', '... Institute') within the paper summaries to add a spatial/organizational dimension to our literature map.

```python
import re

def extract_institutions(text):
    # Basic regex pattern for common institution names
    pattern = r'([A-Z][a-z]+ (?:University|Institute|Laboratory|College|Centre|Center)(?: of [A-Z][a-z]+)?)'
    matches = re.findall(pattern, text)
    return list(set(matches))

df_papers['extracted_institutions'] = df_papers['summary'].apply(extract_institutions)

# Update Graph with Institutions
for _, row in df_papers.iterrows():
    for inst in row['extracted_institutions']:
        G.add_node(inst, label=inst, title=f'Institution: {inst}', color='#dda0dd', group='Institution')
        G.add_edge(row['title'], inst)

# Final comprehensive render
net_final = Network(height='600px', width='100%', bgcolor='#222222', font_color='white', notebook=True, cdn_resources='remote')
net_final.from_nx(G)

filename_final = 'quantum_map_complete.html'
net_final.write_html(filename_final)

with open(filename_final, 'r', encoding='utf-8') as f:
    html_string = f.read()
b64_html = base64.b64encode(html_string.encode('utf-8')).decode('utf-8')
display(HTML(f'<iframe src="data:text/html;base64,{b64_html}" width="100%" height="600px" style="border:none;"></iframe>'))
```
